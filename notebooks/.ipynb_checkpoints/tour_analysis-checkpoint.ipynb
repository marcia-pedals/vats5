```python
import json

with open("/tmp/tour_analysis_parts.json") as f:
    parts = json.load(f)

def fmt_time(seconds):
    neg = seconds < 0
    s = abs(seconds)
    h, s = divmod(s, 3600)
    m, s = divmod(s, 60)
    return f"{'-' if neg else ''}{h:02d}:{m:02d}:{s:02d}"

for p in parts:
    print(f"{p['stop_name']:30s}  relaxed={fmt_time(p['t_relaxed'])}  actual={fmt_time(p['t_actual'])}")
```
